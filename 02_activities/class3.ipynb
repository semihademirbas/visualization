```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1)
ax.plot(x,y2)
fig.show()
```

adding a legend

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1" )
ax.plot(x,y2, label = "Person 2" )
ax.legend(loc='lower right')
fig.show()
```

Modifying the legend

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
       frameon = True, #add frame around the legend
       fontsize = 12, #change font size
       ncol = 2, #specify number of columns
       shadow = True )
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(#loc= 'upper left',
        loc= 'upper left', #specifies the corner of the legend box
       bbox_to_anchor =(1, 1)) 
fig.show()
```

Annotations

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!")
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!",
     ha='center',    ##alignment of text
     color = 'red',  ##modify font colour
     size = 20)      ##modify font size
fig.show()
```

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])

plt.style.use('seaborn-v0_8-bright')
fig, (ax1, ax2) = plt.subplots(ncols=2,
                        nrows=1,
                        figsize=(7, 3))
ax1.scatter(x1,y1)
ax1.xaxis.set_major_locator(plt.MultipleLocator(5))
ax2.bar(x2,y2)
fig.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                               ['ax2', 'ax3']],
                               figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
plt.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                ['ax2', 'ax3']],
                                figsize=(7, 4),
                                layout = 'constrained') 
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18 )
```

Multiple viz on one axes

```python
# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])
# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))
# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "indigo")
ax.plot(x, y2,
color = "red")
```

```python
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax.errorbar(x, #our x values
        y2, #our y values
        yerr = y2_sd,
        fmt = "none")
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2,
       color = "red")
ax.errorbar(x,
       y2,
       yerr = y2_sd,
       fmt = "none",
       ecolor= "indigo",
       elinewidth= 4,
       capsize = 6,
       capthick= 4 )
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2,
       color = "red")
ax.errorbar(x,
       y2,
       yerr = y2_sd,
       fmt = "none",
       ecolor = "indigo",
       elinewidth = 4,
       capsize = 6,
       capthick = 4,
       errorevery= 2 )
```

```python
from PIL import Image # to open images
import requests # to get images from URLs
from io import BytesIO # to store images
```

```python
response = requests.get("https://static.wikia.nocookie.net/the-ultimate-good/images/3/3d/Luffy.png/revision/latest?cb=2023047200546 "
".wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png")
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")

ax_image = fig.add_axes([0.1, # x coordinate (ON FIGURE, NOT AXES)
                       0.11, # y coordinate (ON FIGURE, NOT AXES)
                       0.15, # image width
                       0.35])# image height)

ax_image.imshow(image)
ax_image.axis('off')
```
